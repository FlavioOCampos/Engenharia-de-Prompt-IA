façá jogo de forca em Python

```python
import random

# Lista de palavras
palavras = ["python", "computador", "programacao", "faculdade", "desenvolvimento"]

# Escolhe palavra aleatória
palavra = random.choice(palavras)

letras_descobertas = ["_"] * len(palavra)
tentativas = 6
letras_usadas = []

print("🎮 Jogo da Forca")

while tentativas > 0 and "_" in letras_descobertas:
    print("\nPalavra:", " ".join(letras_descobertas))
    print("Tentativas restantes:", tentativas)
    print("Letras usadas:", ", ".join(letras_usadas))

    letra = input("Digite uma letra: ").lower()

    if letra in letras_usadas:
        print("Você já tentou essa letra!")
        continue

    letras_usadas.append(letra)

    if letra in palavra:
        print("Acertou!")
        for i in range(len(palavra)):
            if palavra[i] == letra:
                letras_descobertas[i] = letra
    else:
        print("Errou!")
        tentativas -= 1

# Resultado final
if "_" not in letras_descobertas:
    print("\nParabéns! Você venceu! 🎉")
else:
    print("\nVocê perdeu 😢 A palavra era:", palavra)
```

refatore esse codigo

```python
import random

class JogoDaForca:
    def __init__(self, palavras):
        self.palavra_secreta = random.choice(palavras).lower()
        self.letras_descobertas = ["_" for _ in self.palavra_secreta]
        self.tentativas_restantes = 6
        self.letras_usadas = set()
        # Nova variável para armazenar as letras únicas da palavra secreta
        # Isso evita recalcular 'set(self.palavra_secreta)' repetidamente.
        self.letras_na_palavra_set = set(self.palavra_secreta)
        # Nova variável para armazenar as letras únicas já descobertas
        # Permite verificar a vitória comparando o tamanho deste conjunto com o 'letras_na_palavra_set'
        self.letras_descobertas_unicas = set()
        self.mensagens = {
            "vitoria": "\nParabéns! Você venceu! 🎉 A palavra era: {self.palavra_secreta}",
            "derrota": "\nVocê perdeu 😢 A palavra era: {self.palavra_secreta}",
            "ja_tentou": "Você já tentou essa letra! Tente outra.",
            "entrada_invalida": "Entrada inválida. Por favor, digite apenas uma letra."
        }

    def _mostrar_estado_do_jogo(self):
        print("\nPalavra: ", " ".join(self.letras_descobertas))
        print(f"Tentativas restantes: {self.tentativas_restantes}")
        print("Letras usadas: ", ", ".join(sorted(list(self.letras_usadas))))

    def _obter_palpite(self):
        while True:
            letra = input("Digite uma letra: ").lower()
            if len(letra) != 1 or not letra.isalpha():
                print(self.mensagens["entrada_invalida"])
            elif letra in self.letras_usadas:
                print(self.mensagens["ja_tentou"])
            else:
                self.letras_usadas.add(letra)
                return letra

    def _verificar_e_atualizar_letra(self, palpite):
        """
        Verifica se a letra está na palavra secreta e atualiza as letras descobertas.
        Retorna True se a letra foi encontrada, False caso contrário.
        """
        letra_encontrada = False
        # Flag para saber se a letra encontrada é uma nova letra única da palavra
        is_new_unique_letter = False

        if palpite in self.letras_na_palavra_set and palpite not in self.letras_descobertas_unicas:
            is_new_unique_letter = True

        for i, char in enumerate(self.palavra_secreta):
            if char == palpite:
                self.letras_descobertas[i] = palpite
                letra_encontrada = True

        # Se a letra foi encontrada e é uma letra única nova para a palavra, adiciona ao conjunto de descobertas
        if letra_encontrada and is_new_unique_letter:
            self.letras_descobertas_unicas.add(palpite)
        return letra_encontrada

    def _processar_palpite(self, palpite):
        if self._verificar_e_atualizar_letra(palpite):
            print("Acertou!")
        else:
            print("Errou!")
            self.tentativas_restantes -= 1

    def jogar(self):
        print("🎮 Jogo da Forca")
        # Refatoração do loop while para melhor performance:
        # A condição de vitória agora compara o número de letras únicas descobertas
        # com o número total de letras únicas na palavra secreta (O(1) operação de conjunto).
        # Isso é mais eficiente do que verificar '"_" in self.letras_descobertas' (O(N) operação de lista/string).
        while self.tentativas_restantes > 0 and \
              len(self.letras_descobertas_unicas) < len(self.letras_na_palavra_set):
            self._mostrar_estado_do_jogo()
            palpite = self._obter_palpite()
            self._processar_palpite(palpite)

        self._mostrar_resultado_final()

    def _mostrar_resultado_final(self):
        # A condição de vitória também é refatorada para usar a comparação de conjuntos
        if len(self.letras_descobertas_unicas) == len(self.letras_na_palavra_set):
            print(self.mensagens["vitoria"].format(self=self))
        else:
            print(self.mensagens["derrota"].format(self=self))

# Lista de palavras para o jogo
palavras_disponiveis = ["python", "computador", "programacao", "faculdade", "desenvolvimento"]

# Criar e iniciar o jogo
jogo = JogoDaForca(palavras_disponiveis)
jogo.jogar()
```
