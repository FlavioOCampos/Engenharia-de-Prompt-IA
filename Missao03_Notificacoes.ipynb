```python
def verificar_status_sensor(leitura):
    """
    Verifica o status de uma leitura do sensor.
    Retorna True se a leitura for crítica (maior que 100), False caso contrário.
    """
    if leitura > 100:
        return True  # Crítico
    else:
        return False # Normal

def disparar_notificacao(mensagem):
    """
    Imprime um alerta visual no console com a mensagem fornecida.
    """
    print("\n" + "*" * 40)
    print(f"!!! ALERTA !!! {mensagem}")
    print("*" * 40 + "\n")
```

### Demonstração de Uso Modular

Com as funções definidas, podemos agora usá-las de forma modular para processar leituras de sensor e disparar notificações quando necessário.

```python
# Exemplos de leituras de sensor
leituras_sensor = [85, 110, 95, 101, 70, 60, 105]

print("Iniciando verificação de leituras de sensor...")

for i, leitura in enumerate(leituras_sensor):
    print(f"\n--- Processando Leitura {i+1} ---")
    print(f"Valor do sensor: {leitura}")

    if verificar_status_sensor(leitura):
        disparar_notificacao(f"Leitura CRÍTICA detectada: {leitura}!")
    else:
        print(f"Leitura {leitura} é normal. Sem alerta.")

print("\nVerificação de leituras concluída.")
```
