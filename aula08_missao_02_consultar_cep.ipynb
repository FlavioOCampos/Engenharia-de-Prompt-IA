```python
import requests

def consultar_cep(cep):

    # URL corrigida para o ViaCEP
    url = f"https://viacep.com.br/ws/{cep}/json/"

    try:
        response = requests.get(url)

        # Verifica se a requisição foi bem-sucedida antes de tentar decodificar o JSON
        response.raise_for_status()
        data = response.json()

        if "erro" in data:
            print("CEP não encontrado.")
        else:
            print(f"Endereço: {data['logradouro']}, {data['bairro']} - {data['localidade']}/{data['uf']}")

    except requests.exceptions.RequestException as e:
        # Captura erros específicos de requisição (conexão, timeout, HTTP etc.)
        print(f"Ocorreu um erro de conexão/requisição: {e}")
    except Exception as e:
        # Captura outros erros inesperados
        print(f"Ocorreu um erro inesperado: {e}")


consultar_cep("01001000")
```
